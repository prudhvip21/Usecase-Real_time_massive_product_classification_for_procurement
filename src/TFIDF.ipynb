```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
import re
from nltk.tokenize import word_tokenize
from nltk import pos_tag
import spacy
```

```python
product_data = pd.read_excel("Test Data 2.xlsx") 
category_data=pd.read_excel("full unspsc list.xlsx")
```

```python
product_data.head()
```

```python
product_data.isna().sum()
```

```python
product_data.dropna(subset=['Short Text'],inplace=True)
```

```python
category_data.head()
```

```python
def preprocess_text(text):
    text=text.lower()
    text=re.sub(r'[^a-zA-Z0-9\s]', '', text)
    return text
```

```python
product_data['Short Text'] = product_data['Short Text'].apply(preprocess_text)
category_data['Level 4 Description'] = category_data['Level 4 Description'].apply(preprocess_text)
```

```python
product_data['Short Text']
```

```python
tfidf = TfidfVectorizer()
```

```python
tfidf.fit(category_data['Level 4 Description'])
```

```python
product_vectors = tfidf.transform(product_data['Short Text'])
category_vectors = tfidf.transform(category_data['Level 4 Description'])
```

```python
for vector in category_vectors.toarray():
    if vector.any():
        print(vector)
```

```python
similarity_matrix = cosine_similarity(product_vectors, category_vectors)
```

```python
result=product_data.copy()
```

```python
result['Mapped Category'] = [
    category_data['Level 4 Description'][similarity_matrix[i].argmax()]
    for i in range(len(product_data))
]
```
