```python
item  = items[120] # pick an item
print('item is', item)
item_encoding = model.encode(item).reshape(1,-1)


level1_similarities = model.similarity(item_encoding, level1_categories_embeddings)

level2_similarities = model.similarity(item_encoding, level2_categories_embeddings)

level3_similarities = model.similarity(item_encoding, level3_categories_embeddings)

level4_similarities = model.similarity(item_encoding, level4_categories_embeddings)
```

```python
level4_similarities = np.array(level4_similarities)

top_50_indices = np.argsort(level4_similarities)[0,-50:][::-1]
print('closest items are')
for i in top_50_indices:
    
    a = float(level1_similarities[0,level1_categories.index(cats.iloc[i]['Level 1 Description'])])
    b = float(level2_similarities[0,level2_categories.index(cats.iloc[i]['.'])]) 
    c = float(level3_similarities[0,level3_categories.index(cats.iloc[i]['Level 3 Description'])] )
    
    print(level4_categories[i],level4_similarities[0,i], round(a+b+c,2) )
```
